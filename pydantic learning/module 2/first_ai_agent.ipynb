## DIFFERENT WAYS OF RUNNING AGENT

### RUN SYNC METHOD

```python
summarizer_agent = Agent(
  "groq:openai/gpt-oss-120b",
  system_prompt ="Summarize the given text into exactly 3 words."
)

summarizer_response = summarizer_agent.run_sync(
  "The quick brown fox jumps over the lazy dog. The dog was not amused by the fox's antics and decided to take a nap instead."
)

print("Sync Output:", summarizer_response.output)
```

### RUN ASYNC METHOD

```python
translator_agent = Agent(
  "groq:openai/gpt-oss-20b",
  system_prompt = "Translate the given text to Hindi"
)
```

```python
translator_agent
```

```python
async def get_async_translation():
  print("Sending request to AI...")
  
  response = await translator_agent.run("Hello, what is your name?")
  print("Async Translation Output:", response.output)
  
await get_async_translation()
```

## RUN STREAM

GEMINI 

```python
story_agent= Agent(
  "groq:openai/gpt-oss-120b",
  system_prompt="Write a creative 2-sentence story"
)
```

```python
async def stream_story():
  print("Streaming response:\n")
  
  async with story_agent.run_stream("Tell me about a brave Knight.") as result:
    async for word_chunk in result.stream_text(delta=True):
      print(word_chunk, end="", flush=True)
      
  print()
    
await stream_story()
```

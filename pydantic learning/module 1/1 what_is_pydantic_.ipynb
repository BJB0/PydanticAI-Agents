## WHAT IS PYDANTIC?

without pydantic

```python
def play_game(score: int):
  print(f"Your current score is: {score}")
  
play_game(10)
play_game("one sixty!")
```

with pydantic

```python
from pydantic import BaseModel

class player(BaseModel):
  name: str
  score: int
  
player1 = player(name="Bhargab", score=90)
print("Player1:", player1)
```

```python
player2 = player(name="sam", score="90")
print("Player2:", player2)
```

```python
player3 = player(name="karpathy", score="ninety")
print("Player3:", player3)
```

## MAKING ROBUST SCHEMAS

```python
from pydantic import BaseModel, Field, field_validator

class UserSignUp(BaseModel):
  username: str = Field(min_length= 3, max_length= 20)
  age: int = Field(ge= 14)
  account_type: str = "free"
  
  @field_validator('username')
  @classmethod
  def check_username_no_spaces(cls, value: str) -> str:
    if " " in value:
      raise ValueError("Username cannot contain spaces!")
    return value

```

GOOD EXAMPLE (NO BREAKING OF BUILT-IN RULES)

```python
user1 = UserSignUp(
  username= "bhargab_jb",
  age= 21
)
print(user1)
```
